### Step 2: Upload Your Reference Audio Sample
Upload `daniel_steady_broadcaster.mp3` or `aravinda_dark_fiction_narrator.mp3` using the file uploader below (or upload via the Colab files sidebar on the left).

```python
from google.colab import files
import os

print("Please upload your reference audio clip (.mp3 or .wav):")
uploaded = files.upload()
ref_audio_file = list(uploaded.keys())[0]
print(f"Uploaded reference voice: {ref_audio_file}")
```

### Step 3: Enter Your Target Script & Clone Voice
Type the new text you want the cloned voice to speak:

```python
from f5_tts.api import F5TTS
from IPython.display import Audio, display

# Enter your new script here
target_script = """Hello everyone, this voice was completely cloned from my ElevenLabs audio sample without paying any subscription fees!"""

print("Loading F5-TTS Zero-Shot model onto GPU...")
f5tts = F5TTS()

print("Generating cloned speech...")
wav, sr, spect = f5tts.infer(
    ref_file=ref_audio_file,
    ref_text="",  # Optional: transcription of reference clip if known
    gen_text=target_script
)

output_file = "cloned_voice_output.wav"
import soundfile as sf
sf.write(output_file, wav, sr)
print(f"Audio saved to {output_file}!")

# Play the cloned audio inline
display(Audio(output_file))
```
